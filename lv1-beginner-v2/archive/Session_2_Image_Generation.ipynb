## 🎯 The Big Picture

### How AI Creates Images

```
📝 Your Text Prompt    →    🧠 Diffusion Model    →    🎨 Generated Image
"a cat wearing a hat"        (noise → image)          [beautiful image!]
```

AI image generation uses **Diffusion Models** - they start with random noise and gradually "paint" your image based on your text description.

<details>
<summary><strong>🔬 How Do Diffusion Models Work? (Click to Expand)</strong></summary>

### The Diffusion Process

1. **Forward Process (Training)**: Add noise to images until they're pure static
2. **Reverse Process (Generation)**: Learn to remove noise step by step

```
Step 1: 📺 Pure noise (random pixels)
Step 2: 🌫️ Some shapes emerge
Step 3: 🎭 Details appear
Step 4: 🎨 Final image!
```

The text prompt "guides" which direction to denoise → creating matching images!

</details>

````python
def generate_image(prompt, size="1024x1024", quality="standard", save_as=None):
    """Generate an image using DALL·E 3.

    Args:
        prompt: Text description of the image
        size: "1024x1024", "1792x1024" (wide), or "1024x1792" (tall)
        quality: "standard" or "hd" (more detail, costs more)
        save_as: Optional filename to save the image (e.g. "hero.png")

    Returns:
        PIL Image object
    """
    response = client.images.generate(
        model="dall-e-3",
        prompt=prompt,
        size=size,
        quality=quality,
        n=1
    )

    url = response.data[0].url
    img_data = requests.get(url).content
    image = Image.open(BytesIO(img_data))

    if save_as:
        os.makedirs("game_assets", exist_ok=True)
        image.save(f"game_assets/{save_as}")
        print(f"💾 Saved: game_assets/{save_as}")

    return image


def show_image(image, title=""):
    """Display a PIL image inline in the notebook."""
    plt.figure(figsize=(6, 6))
    plt.imshow(image)
    plt.axis("off")
    if title:
        plt.title(title, fontsize=12)
    plt.tight_layout()
    plt.show()


print("🎨 generate_image() and show_image() are ready!")
````

## ✍️ Part 3: Advanced Prompting

### The Prompt Recipe for Images

```
SUBJECT + STYLE + QUALITY + LIGHTING + DETAILS + NEGATIVE
```

| Component | Example | Effect |
|:--|:--|:--|
| **Subject** | "a wise owl wizard" | What to generate |
| **Style** | "studio ghibli style" | Artistic look |
| **Quality** | "8k, highly detailed" | Resolution/detail |
| **Lighting** | "golden hour lighting" | Mood/atmosphere |
| **Details** | "wearing spectacles" | Specific features |
| **Negative** | "blurry, distorted" | What to avoid |

````python
# Comparison: Simple vs Detailed prompts

# Simple prompt
simple_prompt = "a dragon"

# Detailed prompt
detailed_prompt = """a majestic dragon with iridescent scales,
perched on a crystal mountain peak, fantasy art style,
dramatic lighting, highly detailed, 8k resolution,
chromatic aberration, volumetric fog"""

print("📌 Simple prompt:")
image1 = generate_image(simple_prompt)
show_image(image1, "Simple: 'a dragon'")

print("📌 Detailed prompt:")
image2 = generate_image(detailed_prompt)
show_image(image2, "Detailed prompt")
````

### 🎭 Style Keywords Reference

| Category | Keywords |
|:--|:--|
| **Art Styles** | oil painting, watercolor, pixel art, anime, photorealistic, pencil sketch |
| **Lighting** | golden hour, dramatic lighting, neon glow, soft lighting, backlit |
| **Quality** | 8k, highly detailed, professional, masterpiece, award-winning |
| **Mood** | serene, dramatic, mysterious, cheerful, dark, whimsical |
| **Visual Effects** | bokeh, depth of field, lens flare, motion blur, chromatic aberration |

````python
# Same subject, different styles!

subject = "a cat sitting on a windowsill"
styles = [
    "anime style, studio ghibli",
    "oil painting, classical, rembrandt lighting",
    "pixel art, 16-bit, retro game style",
    "cyberpunk, neon lights, futuristic"
]

print("🎨 Same subject, different styles:\n")
for style in styles:
    prompt = f"{subject}, {style}"
    print(f"Style: {style}")
    image = generate_image(prompt)
    show_image(image, style)
````

## 🧩 Part 4: Negative Prompts (Optional)

**Negative prompts** tell the AI what NOT to include.

````python
def generate_with_style(prompt, style_suffix="", size="1024x1024"):
    """Generate image with an extra style instruction appended."""
    full_prompt = f"{prompt}, {style_suffix}" if style_suffix else prompt
    print(f"📝 Prompt: {full_prompt}")
    return generate_image(full_prompt, size=size)


# With DALL·E 3, negative concepts go *in* the prompt itself:
# e.g. 'a friendly robot, clean background, bright colors, no damage, no darkness'
prompt = "a portrait of a friendly robot, digital art, bright colors, clean and undamaged"
image = generate_with_style(prompt)
show_image(image, "Friendly Robot")
````

### Common Negative Prompts

| For | Negative Words |
|:--|:--|
| **Quality** | blurry, low quality, distorted, ugly, deformed |
| **People** | extra fingers, bad anatomy, cloned face |
| **Style** | cartoon (if you want realism), plain, boring |
| **Content** | violent, scary, nsfw |
