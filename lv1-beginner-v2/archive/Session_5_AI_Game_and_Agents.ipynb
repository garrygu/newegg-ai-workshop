## 🎮 Part 2: Basic Guessing Game

```python
class AIGuessingGame:
    """An AI-powered guessing game with sentiment awareness."""
    
    def __init__(self):
        self.categories = ['cat', 'dog', 'car', 'airplane', 'bird', 
                          'ship', 'truck', 'horse', 'frog', 'deer']
        self.score = 0
        self.rounds = 0
        self.secret = None
        self.analyzer = sentiment_analyzer
    
    def new_round(self):
        """Start a new round with a random secret."""
        self.secret = random.choice(self.categories)
        self.rounds += 1
        print(f"\n🎮 Round {self.rounds}!")
        print("🤫 I'm thinking of something...")
        print("💡 Give me clues and I'll try to guess!")
    
    def get_hint(self):
        """Generate a hint based on the secret."""
        hints = {
            'cat': "It purrs and likes to sleep.",
            'dog': "It barks and loves to play fetch.",
            'car': "It has four wheels and you drive it.",
            'airplane': "It flies high in the sky.",
            'bird': "It has wings and can fly.",
            'ship': "It sails on the ocean.",
            'truck': "It's big and carries heavy loads.",
            'horse': "You can ride it and it says neigh.",
            'frog': "It hops and says ribbit.",
            'deer': "It has antlers and lives in forests."
        }
        return hints.get(self.secret, "It's something interesting!")
    
    def process_clue(self, clue):
        """Analyze clue sentiment and make a guess."""
        # Analyze sentiment
        result = self.analyzer(clue)[0]
        sentiment = result['label']
        
        # React to sentiment
        if sentiment == 'POSITIVE':
            print("😊 You seem excited! That's a good clue!")
        else:
            print("🤔 Hmm, interesting clue...")
        
        # Make a guess (simple keyword matching for now)
        clue_lower = clue.lower()
        for category in self.categories:
            if category in clue_lower:
                return category
        
        # Random guess if no match
        return random.choice(self.categories)
    
    def check_guess(self, guess):
        """Check if the AI's guess is correct."""
        if guess == self.secret:
            self.score += 1
            print(f"\n🎉 I got it! It's a {guess}!")
            print(f"🏆 Score: {self.score}/{self.rounds}")
            return True
        else:
            print(f"\n🤖 I guess... a {guess}?")
            return False
    
    def play_round(self):
        """Play one complete round."""
        self.new_round()
        print(f"\n💭 Hint: {self.get_hint()}")
        
        attempts = 3
        for i in range(attempts):
            clue = input(f"\nClue {i+1}/{attempts}: ")
            guess = self.process_clue(clue)
            
            if self.check_guess(guess):
                return True
            elif i < attempts - 1:
                print("Try another clue!")
        
        print(f"\n😅 The answer was: {self.secret}")
        return False
```

```python
# Quick test
game = AIGuessingGame()
print("🎮 AI Guessing Game initialized!")
print(f"📚 Categories: {', '.join(game.categories)}")
```

## 🔗 Part 3: Enhanced Game with All Components

```python
class AdvancedAIGame:
    """
    Advanced AI Game that combines:
    - Sentiment analysis (emotional responses)
    - LLM integration (smart hints)
    - Image generation (visual clues)
    """
    
    def __init__(self, name="AI Game Host"):
        self.name = name
        self.score = 0
        self.total_rounds = 0
        self.analyzer = sentiment_analyzer
        
        # Game content
        self.items = [
            {"name": "robot", "hint": "I'm made of metal and beep boop!", "emoji": "🤖"},
            {"name": "dragon", "hint": "I breathe fire and have wings!", "emoji": "🐉"},
            {"name": "spaceship", "hint": "I fly through the stars!", "emoji": "🚀"},
            {"name": "wizard", "hint": "I cast magical spells!", "emoji": "🧙"},
            {"name": "unicorn", "hint": "I'm a horse with a magical horn!", "emoji": "🦄"},
        ]
        
        # Emotional responses
        self.positive_reactions = [
            "🌟 Your enthusiasm is amazing!",
            "💪 You're doing great!",
            "🎯 Getting warmer!"
        ]
        self.negative_reactions = [
            "🤔 Don't give up!",
            "💡 Here's another hint...",
            "🌈 Keep trying, you've got this!"
        ]
    
    def welcome(self):
        """Display welcome message."""
        print("\n" + "="*50)
        print(f"    🎮 Welcome to {self.name}! 🎮")
        print("="*50)
        print("\nI'll give you hints about something I'm thinking of.")
        print("Try to guess what it is!\n")
    
    def react_to_sentiment(self, text):
        """React based on player's emotional tone."""
        result = self.analyzer(text)[0]
        if result['label'] == 'POSITIVE':
            return random.choice(self.positive_reactions)
        else:
            return random.choice(self.negative_reactions)
    
    def play_round(self, item):
        """Play one round with a specific item."""
        print(f"\n🎲 Round {self.total_rounds + 1}")
        print("-" * 30)
        print(f"💭 Hint: {item['hint']}")
        
        for attempt in range(3):
            guess = input(f"\n❓ Your guess (attempt {attempt+1}/3): ").strip().lower()
            
            # React to how they're feeling
            reaction = self.react_to_sentiment(guess)
            print(reaction)
            
            if item['name'] in guess:
                self.score += 1
                print(f"\n🎉 CORRECT! It was a {item['emoji']} {item['name']}!")
                print(f"🏆 Score: {self.score}")
                return True
            else:
                if attempt < 2:
                    print("Not quite! Try again...")
        
        print(f"\n😢 It was a {item['emoji']} {item['name']}!")
        return False
    
    def play(self, num_rounds=3):
        """Play the complete game."""
        self.welcome()
        
        # Shuffle and pick items
        game_items = random.sample(self.items, min(num_rounds, len(self.items)))
        
        for item in game_items:
            self.total_rounds += 1
            self.play_round(item)
        
        # Final score
        print("\n" + "="*50)
        print("    🏆 GAME OVER! 🏆")
        print("="*50)
        print(f"\n📊 Final Score: {self.score}/{self.total_rounds}")
        
        if self.score == self.total_rounds:
            print("🌟 PERFECT SCORE! You're amazing!")
        elif self.score >= self.total_rounds // 2:
            print("🎯 Great job! You're getting good at this!")
        else:
            print("💪 Keep practicing! You'll get better!")
```

```python
# Create the advanced game
advanced_game = AdvancedAIGame("Super AI Challenge")
print("🎮 Advanced game ready!")
```

### What is an AI Agent?

An **AI Agent** is an AI that can:
- 🧠 **Think** - Reason about what to do
- 🛠️ **Use Tools** - Call functions/APIs
- 🔄 **Loop** - Keep going until task is done

```python
class SimpleAgent:
    """A simple AI agent that can use tools."""
    
    def __init__(self):
        self.tools = {
            "analyze_sentiment": self.analyze_sentiment,
            "generate_hint": self.generate_hint,
            "get_random_item": self.get_random_item,
        }
    
    def analyze_sentiment(self, text):
        """Tool: Analyze text sentiment."""
        result = sentiment_analyzer(text)[0]
        return f"Sentiment: {result['label']} ({result['score']:.1%})"
    
    def generate_hint(self, item):
        """Tool: Generate a hint for an item."""
        hints = {
            "robot": "It's made of metal and circuits!",
            "dragon": "It breathes fire!",
            "default": "It's something interesting!"
        }
        return hints.get(item, hints["default"])
    
    def get_random_item(self, category=None):
        """Tool: Get a random game item."""
        items = ["robot", "dragon", "spaceship", "wizard"]
        return random.choice(items)
    
    def run(self, task):
        """Execute a task using available tools."""
        print(f"\n🤖 Agent thinking about: {task}")
        print("📋 Available tools:", list(self.tools.keys()))
        
        # Simple reasoning (in real agents, this would be LLM-powered)
        if "sentiment" in task.lower():
            tool = "analyze_sentiment"
            arg = task.split(":")[-1].strip() if ":" in task else "I love this!"
        elif "hint" in task.lower():
            tool = "generate_hint"
            arg = "robot"
        else:
            tool = "get_random_item"
            arg = None
        
        print(f"🛠️ Using tool: {tool}")
        result = self.tools[tool](arg) if arg else self.tools[tool]()
        print(f"📤 Result: {result}")
        return result
```

```python
# Test the agent
agent = SimpleAgent()

# Try different tasks
agent.run("analyze sentiment: I'm so excited about AI!")
agent.run("generate a hint for the game")
agent.run("pick a random item")
```
