## Step 1 — Load Settings from Session 3

```python
import json
import random
from pathlib import Path

settings_path = Path("session_data/game_settings.json")

if settings_path.exists():
    with open(settings_path, "r", encoding="utf-8") as f:
        settings = json.load(f)
else:
    settings = {"classes": ["cat", "dog"], "num_classes": 2, "personality": "friendly"}

classes = settings["classes"]
print("Loaded labels:", classes)
print("Current personality:", settings.get("personality", "friendly"))
```

## Step 2 — Choose a Personality

Try one of these:
- `"friendly"`
- `"funny"`
- `"coach"`

```python
personality = "friendly"   # Change me!
print("Chosen personality:", personality)
```

## Step 3 — Create Response Lists

These are phrases your AI can say.

```python
response_bank = {
    "friendly": {
        "correct": [
            "Great job! 🎉",
            "Nice work! You got it!",
            "Awesome! Keep going!"
        ],
        "wrong": [
            "Good try! Let's do another one.",
            "Not quite, but you're learning!",
            "That was a brave guess!"
        ]
    },
    "funny": {
        "correct": [
            "Boom! You cracked it! 😆",
            "That guess was chef's kiss. 👨‍🍳",
            "Your brain is on fire! 🔥"
        ],
        "wrong": [
            "Oopsie! The AI is giggling. 😂",
            "Nope—but that guess had style!",
            "Plot twist! Try again!"
        ]
    },
    "coach": {
        "correct": [
            "Excellent focus. Keep it up! 💪",
            "Strong work. Ready for the next round?",
            "You are improving fast!"
        ],
        "wrong": [
            "That’s okay. Mistakes help us learn.",
            "Reset and try the next one.",
            "Stay calm—you’ve got this."
        ]
    }
}
```

## Step 4 — Build a Function for Game Responses

```python
def game_response(correct, personality_name):
    mode = "correct" if correct else "wrong"
    return random.choice(response_bank[personality_name][mode])

print(game_response(True, personality))
print(game_response(False, personality))
```

## Step 5 — Test Different Situations

Run this cell a few times.

```python
for result in [True, False, False, True]:
    print("Correct?" , result, "->", game_response(result, personality))
```

## Step 6 — Add a Hint Message

Hints help the player without giving away the full answer.

```python
hint_bank = {
    "cat": "Hint: This animal often says meow. 🐱",
    "dog": "Hint: This animal often says woof. 🐶",
}

def hint_for_label(label):
    return hint_bank.get(label, "Hint: Look carefully at the picture.")

for label in classes:
    print(label, "->", hint_for_label(label))
```

## Step 7 — Create a Welcome Message

```python
def welcome_message(personality_name):
    if personality_name == "friendly":
        return "Hi! I'm your AI buddy. Let's play! 😊"
    if personality_name == "funny":
        return "Welcome, human! Prepare for maximum guessing chaos! 🤖"
    if personality_name == "coach":
        return "Welcome back. Stay focused and do your best. 💪"

print(welcome_message(personality))
```

## Step 8 — Try Your Own Custom Messages

Edit the lists or the welcome message to make your AI feel unique.

```python
# Add one custom message here
response_bank[personality]["correct"].append("Yes! That was a smart guess!")
response_bank[personality]["wrong"].append("No worries—every guess teaches us something.")

print("Custom responses added.")
print(game_response(True, personality))
print(game_response(False, personality))
```

## Step 9 — Save Settings for Session 5

```python
settings["personality"] = personality
settings["response_bank"] = response_bank
settings["hint_bank"] = hint_bank

Path("session_data").mkdir(exist_ok=True)
with open(settings_path, "w", encoding="utf-8") as f:
    json.dump(settings, f, indent=2)

print("✅ Saved updated settings for the final game.")
```
