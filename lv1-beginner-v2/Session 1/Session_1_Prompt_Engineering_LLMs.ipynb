## 📚 Part 1: What Are LLMs?

**LLM = Large Language Model**

An LLM is a type of AI trained on a huge amount of text — including books, websites, articles, and code.

It learns patterns in language and uses those patterns to predict what word or phrase should come next.

> 💡 You can think of it as a **super-powerful autocomplete**.

That is why it can do things like:
- ✅ Answer questions
- ✅ Summarize text
- ✅ Translate languages
- ✅ Write stories
- ✅ Help with coding

### How it works at a simple level

```
📚 Lots of Text  →  🧠 Learns Patterns  →  📝 Predicts Next Words
```

<details>
<summary><strong>🔬 More Technical Version (Click to Expand)</strong></summary>

```
📚 Training Data  →  🧠 Neural Network  →  📝 Text Prediction
(Books, websites,      (Billions of            (Generates text one token
 code, articles)       parameters)              a time)
```

LLMs use a technology called **Transformers** (invented in 2017):

1. **Tokenization**: Text is split into small pieces called tokens
   - "Hello, how are you?" → ["Hello", ",", " how", " are", " you", "?"]

2. **Attention**: The model looks at how tokens relate to each other
   - "The cat sat on the mat" → "cat" pays attention to "sat" and "mat"

3. **Prediction**: Based on patterns, predicts the next most likely token
   - Input: "The sky is" → Prediction: "blue" (most likely)

### Key Insight
LLMs don't "understand" like humans — they generate likely text based on patterns learned from training data.

</details>

## ✍️ Part 2: Prompt Engineering

**Prompt Engineering** = the skill of writing clear and effective prompts to get better results from AI.

This is one of the most important AI skills you can learn!

### 🎯 Prompt Engineering Techniques

#### 1️⃣ Zero-Shot Prompting

Ask directly without giving examples.

```text
Prompt: "What is the capital of France?"
Output: "The capital of France is Paris."
```

#### 2️⃣ Few-Shot Prompting

Give a few examples to show the pattern you want.

```text
Prompt:
"Classify the sentiment:
Text: 'I love this!' → Positive
Text: 'This is terrible' → Negative
Text: 'Best day ever!' →"

Output: "Positive"
```

#### 3️⃣ Step-by-Step Prompting

Ask AI to explain the solution step by step.

```text
Prompt: "What is 17 × 23? Solve it step by step."
Output:
"17 × 20 = 340
17 × 3 = 51
340 + 51 = 391
Answer: 391"
```

> This technique is often called **Chain-of-Thought prompting**.

### 🧩 The Prompt Recipe

```text
┌─────────────────────────────────────────────────────────────────────┐
│                         PROMPT RECIPE                              │
├─────────────────────────────────────────────────────────────────────┤
│  1. ROLE         → Set the role ("You are a helpful tutor")        │
│  2. CONTEXT      → Give background information                     │
│  3. TASK         → Say what you want AI to do                      │
│  4. FORMAT       → Describe the output format                      │
│                    (list, paragraph, table, code)                  │
│  5. EXAMPLES     → Show a pattern if helpful                       │
│  6. CONSTRAINTS  → Add limits or requirements                      │
│                    (simple English, 3 bullets, under 100 words)    │
└─────────────────────────────────────────────────────────────────────┘
```

> You do not always need all 6 parts.
> But clearer prompts usually lead to better answers.

### ❌ Bad vs ✅ Good Prompts

| Problem  | ❌ Bad Prompt       | ✅ Good Prompt                                                                                                                                                          | Why the good prompt works                                                                                                |
| :------- | :----------------- | :--------------------------------------------------------------------------------------------------------------------------------------------------------------------- | :----------------------------------------------------------------------------------------------------------------------- |
| Writing  | "Write about dogs" | "Write a 200-word blog post about the benefits of adopting shelter dogs for first-time pet owners. Include 3 practical tips."                                          | It clearly defines the **topic**, **audience**, **length**, and **what to include**.                                     |
| Coding   | "Fix my code"      | "You are a Python expert. Debug the code below. It is supposed to calculate factorial, but it gives the wrong output. Explain the problem and provide corrected code." | It gives the AI a **role**, explains the **task**, describes the **problem**, and asks for a specific **output format**. |
| Learning | "Help me"          | "I'm a high school student learning calculus. Explain derivatives in simple language using one real-world example I can relate to."                                    | It provides **student context**, defines the **topic**, and sets the **difficulty level** and **teaching style**.        |

### Key Takeaway

> Better prompts give AI more direction.

Or in a slightly more structured way:

> Strong prompts usually include:
>
> * clear task
> * useful context
> * target audience or role
> * desired format or constraints

## 🤖 Part 4: Your First API Call

Now let's send our first prompt to the AI!

We’ll create a simple function that sends a prompt and returns the response.

`````python
def ask_ai(prompt, model="gpt-4o-mini"):
    """Send a prompt to the AI and return the response."""

    response = client.responses.create(
        model=model,
        input=prompt
    )

    return response.output[0].content[0].text


# Test it!
result = ask_ai("What is AI in one sentence?")
print("🤖 AI says:", result)
`````

## 🎭 Part 5: System Prompts & Personas

**System prompts** let you control how the AI behaves — like giving it a personality or role.

Instead of just asking a question, you can tell the AI:
- who it is
- how it should respond
- what tone or style to use

This is called creating a **persona**.

`````python
def ask_with_persona(system_prompt, user_prompt, model="gpt-4o-mini"):
    """Ask AI with a specific persona/system prompt."""
    try:
        response = client.chat.completions.create(
            model=model,
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ]
        )
        return response.choices[0].message.content

    except Exception as e:
        return f"❌ Error: {e}"
`````

<details>
<summary><strong>📝 Why do we use a system prompt?</strong></summary>

- The system prompt defines the AI's personality and behavior
- It influences *how* the AI responds, not just *what* it says
- It stays consistent throughout the conversation

> In this section, we use the chat format because it allows us to define roles like "system" and "user".
</details>

## 🌡️ Part 6: Temperature & Creativity

**Temperature** changes how predictable or creative the AI’s response is.

Low temperature → more focused and consistent

High temperature → more varied and creative

`````python
def ask_ai_with_temperature(prompt, temperature=1.0):
    """Send a prompt with a temperature setting to control creativity."""
    try:
        response = client.chat.completions.create(
            model="gpt-4o-mini",
            messages=[{"role": "user", "content": prompt}],
            temperature=temperature  # 0.0 = focused, ~1.0 = balanced, >1.0 = more creative
        )
        return response.choices[0].message.content

    except Exception as e:
        return f"❌ Error: {e}"
`````

### 🔍 What to look for

Compare the outputs:
- Which one is the most predictable?
- Which one is the most creative?
- Which one do you prefer?

Notice how the same prompt produces different styles of answers.

`````python
prompt = "Write a one-sentence imaginative story about a robot."

low = ask_ai_with_temperature(prompt, temperature=0.0)
mid = ask_ai_with_temperature(prompt, temperature=1.0)
high = ask_ai_with_temperature(prompt, temperature=1.5)

print("📝 Prompt:", prompt)
print()

print("🧊 Temperature 0.0 (Focused/Predictable):")
print(low)
print()

print("🌡️ Temperature 1.0 (Balanced/Default):")
print(mid)
print()

print("🔥 Temperature 1.5 (Creative/Wild):")
print(high)
`````

### 🌡️ Temperature Guidelines

| Temperature | Behavior | Best For | Examples |
|:--:|:--|:--|:--|
| 0.0 – 0.3 | Very consistent, predictable | Factual tasks | Code, math, definitions |
| 0.4 – 0.7 | Balanced, slightly varied | General tasks | Explanations, summaries |
| 0.8 – 1.2 | Creative, more diverse | Creative tasks | Stories, brainstorming |
| 1.3 – 2.0 | Highly random, unpredictable | Experimental | Poetry, wild ideas |

💡 **Rule of thumb:**
- If you want accuracy → use lower temperature  
- If you want creativity → use higher temperature  

⚠️ **Note:**
Very high temperatures (>1.3) can produce less accurate or less coherent answers.

## 🏗️ Part 7: Build Your AI Assistant!

Let’s build a simple AI assistant with **conversation memory**.

That means the assistant will remember earlier messages in this chat, so it can respond more naturally and stay on topic.

### How conversation memory works

We store every message in a list called `conversation_history`.

Each time the user sends a new message, we send the whole conversation history to the AI again.
That is how the assistant “remembers” earlier parts of the chat.

`````python
class AIAssistant:
    """A simple AI assistant with conversation memory."""

    def __init__(self, name, system_prompt):
        self.name = name
        self.system_prompt = system_prompt
        self.conversation_history = [
            {"role": "system", "content": system_prompt}
        ]

    def chat(self, user_message):
        """Send a message and get a response."""
        user_entry = {"role": "user", "content": user_message}

        try:
            # Add user message to history
            self.conversation_history.append(user_entry)

            # Get AI response
            response = client.chat.completions.create(
                model="gpt-4o-mini",
                messages=self.conversation_history
            )

            assistant_message = response.choices[0].message.content

            # Add assistant message to history
            self.conversation_history.append(
                {"role": "assistant", "content": assistant_message}
            )

            return assistant_message

        except Exception as e:
            # Remove the last user message if the request failed
            if self.conversation_history and self.conversation_history[-1] == user_entry:
                self.conversation_history.pop()
            return f"❌ Error: {e}"

    def reset(self):
        """Clear conversation history."""
        self.conversation_history = [
            {"role": "system", "content": self.system_prompt}
        ]
        print(f"🔄 {self.name} memory cleared!")
`````

`````python
# Create your assistant!
homework_helper = AIAssistant(
    name="Study Buddy",
    system_prompt="""
You are Study Buddy, a friendly AI assistant for high school students.

Rules:
- Explain concepts clearly without simply giving away the final answer
- Use relatable examples
- Ask guiding questions to help the student think
- Encourage effort and celebrate progress
- Keep responses concise and easy to understand
"""
)

print("🎒 Study Buddy is ready to help!")
`````

### 🎮 Try it yourself: Chat with your AI assistant

Now you can talk to your AI assistant just like ChatGPT!

Try:
- asking follow-up questions
- testing its memory
- changing the topic
- resetting the conversation

`````python
# Interactive chat loop
def interactive_chat(assistant):
    """Start an interactive chat session."""
    print(f"\n🤖 Chat with {assistant.name}!")
    print("Type 'quit' to exit, 'reset' to clear memory, 'help' for commands.\n")
    
    while True:
        user_input = input("You: ")
        
        if not user_input.strip():
            continue

        if user_input.lower() == 'quit':
            print("👋 Goodbye!")
            break
        elif user_input.lower() == 'reset':
            assistant.reset()
            continue
        elif user_input.lower() == 'help':
            print("Commands: quit, reset, help")
            continue
        
        try:
            response = assistant.chat(user_input)
            print(f"\n🤖 {assistant.name}: {response}\n")
        except Exception as e:
            print(f"❌ Error: {e}")

# Uncomment to start chatting:
# interactive_chat(homework_helper)
`````
