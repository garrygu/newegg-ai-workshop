### The Transformer Architecture (Simplified)

LLMs use a technology called **Transformers** (invented in 2017):

1. **Tokenization**: Text is split into tokens (words or parts of words)
   - "Hello, how are you?" → ["Hello", ",", " how", " are", " you", "?"]

2. **Attention**: The model looks at relationships between ALL tokens
   - "The cat sat on the mat" → "cat" pays attention to "sat" and "mat"

3. **Prediction**: Based on patterns, predicts the next most likely token
   - Input: "The sky is" → Prediction: "blue" (most likely)

### Key Insight
LLMs don't "understand" like humans. They predict statistically likely text based on patterns in training data!

## ✍️ Part 2: Prompt Engineering

**Prompt Engineering** = The art and science of writing effective prompts to get great AI outputs.

This is the MOST important AI skill you can learn!

### 🎯 Prompt Engineering Techniques

#### 1️⃣ Zero-Shot Prompting
Just ask directly, no examples needed.

```
Prompt: "What is the capital of France?"
Output: "The capital of France is Paris."
```

#### 2️⃣ Few-Shot Prompting
Give examples to show the pattern you want.

```
Prompt: 
"Classify the sentiment:
Text: 'I love this!' → Positive
Text: 'This is terrible' → Negative
Text: 'Best day ever!'"

Output: "Positive"
```

#### 3️⃣ Chain-of-Thought
Ask AI to think step by step.

```
Prompt: "What is 17 × 23? Think step by step."
Output: "Let me break this down:
17 × 20 = 340
17 × 3 = 51
340 + 51 = 391
Answer: 391"
```

### 🧩 The Perfect Prompt Recipe

```
┌─────────────────────────────────────────────────────────────────┐
│                    PERFECT PROMPT FORMULA                       │
├─────────────────────────────────────────────────────────────────┤
│  1. ROLE      → Tell AI who to be ("You are a helpful tutor")  │
│  2. CONTEXT   → Provide background info                        │
│  3. TASK      → What you want AI to do                         │
│  4. FORMAT    → How you want the output (list, essay, code)    │
│  5. EXAMPLES  → Show what good output looks like               │
│  6. CONSTRAINTS → Any limits or requirements                   │
└─────────────────────────────────────────────────────────────────┘
```

### ❌ Bad vs ✅ Good Prompts

| ❌ Bad Prompt | ✅ Good Prompt |
|:--|:--|
| "Write about dogs" | "Write a 200-word blog post about the benefits of adopting shelter dogs, targeting first-time pet owners. Include 3 key tips." |
| "Fix my code" | "You are a Python expert. Debug this code that should calculate factorial but gives wrong output. Explain what's wrong and provide fixed code." |
| "Help me" | "I'm a high school student learning calculus. Explain derivatives using a real-world example I can relate to." |

## 🤖 Part 4: Your First API Call

````python
def ask_ai(prompt, model="gpt-3.5-turbo"):
    """Simple function to ask the AI a question."""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "user", "content": prompt}
        ]
    )
    return response.choices[0].message.content

# Test it!
result = ask_ai("What is AI in one sentence?")
print("🤖 AI says:", result)
````

## 🎭 Part 5: System Prompts & Personas

**System prompts** set the AI's personality and behavior for the entire conversation.

````python
def ask_with_persona(system_prompt, user_prompt, model="gpt-3.5-turbo"):
    """Ask AI with a specific persona/system prompt."""
    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": user_prompt}
        ]
    )
    return response.choices[0].message.content
````

````python
# Example: Pirate persona 🏴‍☠️
pirate_system = """
You are a friendly pirate captain. You speak like a pirate 
(using "arr", "matey", "landlubber", etc.) but you're very 
knowledgeable about science and technology. Keep responses brief.
"""

response = ask_with_persona(pirate_system, "Explain how computers work.")
print("🏴‍☠️ Pirate Captain says:")
print(response)
````

````python
# Example: Encouraging tutor 📚
tutor_system = """
You are an encouraging and patient math tutor for high school 
students. You:
- Use simple language
- Give step-by-step explanations
- Celebrate small wins with emojis
- Never make the student feel dumb
- End with a practice question
"""

response = ask_with_persona(tutor_system, "I don't understand fractions.")
print("📚 Math Tutor says:")
print(response)
````

## 🌡️ Part 6: Temperature & Creativity

**Temperature** controls how creative/random the AI's responses are.

````python
def ask_with_temperature(prompt, temperature=1.0):
    """Ask with a specific temperature setting."""
    response = client.chat.completions.create(
        model="gpt-3.5-turbo",
        messages=[{"role": "user", "content": prompt}],
        temperature=temperature  # 0.0 = focused, 2.0 = very creative
    )
    return response.choices[0].message.content
````

````python
# Compare temperatures!
prompt = "Write a one-sentence story about a robot."

print("🧊 Temperature 0.0 (Focused/Predictable):")
print(ask_with_temperature(prompt, temperature=0.0))
print()

print("🌡️ Temperature 1.0 (Balanced/Default):")
print(ask_with_temperature(prompt, temperature=1.0))
print()

print("🔥 Temperature 1.5 (Creative/Wild):")
print(ask_with_temperature(prompt, temperature=1.5))
````

### Temperature Guidelines

| Temperature | Best For | Examples |
|:--:|:--|:--|
| 0.0 - 0.3 | Factual tasks | Code, math, facts |
| 0.4 - 0.7 | Balanced tasks | Writing, explanations |
| 0.8 - 1.2 | Creative tasks | Stories, brainstorming |
| 1.3 - 2.0 | Very creative | Poetry, wild ideas |

## 🏗️ Part 7: Build Your AI Assistant!

Let's create a complete AI assistant with conversation memory!

````python
class AIAssistant:
    """A simple AI assistant with conversation memory."""
    
    def __init__(self, name, system_prompt):
        self.name = name
        self.system_prompt = system_prompt
        self.conversation_history = [
            {"role": "system", "content": system_prompt}
        ]
    
    def chat(self, user_message):
        """Send a message and get a response."""
        # Add user message to history
        self.conversation_history.append(
            {"role": "user", "content": user_message}
        )
        
        # Get AI response
        response = client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=self.conversation_history
        )
        
        assistant_message = response.choices[0].message.content
        
        # Add assistant message to history
        self.conversation_history.append(
            {"role": "assistant", "content": assistant_message}
        )
        
        return assistant_message
    
    def reset(self):
        """Clear conversation history."""
        self.conversation_history = [
            {"role": "system", "content": self.system_prompt}
        ]
        print(f"🔄 {self.name} memory cleared!")
````

````python
# Create your assistant!
homework_helper = AIAssistant(
    name="Study Buddy",
    system_prompt="""
    You are Study Buddy, a friendly AI assistant that helps high school 
    students with their homework. You:
    - Explain concepts clearly without giving direct answers
    - Use relatable examples
    - Ask guiding questions to help students think
    - Encourage and celebrate progress
    - Keep responses concise (2-3 paragraphs max)
    """
)

print("🎒 Study Buddy is ready to help!")
````

````python
# Test conversation with memory!
print("You: What is a derivative in calculus?")
print(f"\n🤖 Study Buddy: {homework_helper.chat('What is a derivative in calculus?')}")
````

````python
# Notice it remembers the context!
print("You: Can you give me a real-world example?")
print(f"\n🤖 Study Buddy: {homework_helper.chat('Can you give me a real-world example?')}")
````

````python
# Interactive chat loop
def interactive_chat(assistant):
    """Start an interactive chat session."""
    print(f"\n🤖 Chat with {assistant.name}!")
    print("Type 'quit' to exit, 'reset' to clear memory.\n")
    
    while True:
        user_input = input("You: ")
        
        if user_input.lower() == 'quit':
            print("👋 Goodbye!")
            break
        elif user_input.lower() == 'reset':
            assistant.reset()
            continue
        
        response = assistant.chat(user_input)
        print(f"\n🤖 {assistant.name}: {response}\n")

# Uncomment to start chatting:
# interactive_chat(homework_helper)
````
