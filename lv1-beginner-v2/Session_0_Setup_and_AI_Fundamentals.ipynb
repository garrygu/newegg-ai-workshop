## 🐍 Part 2: Python Basics

Python is the #1 language for AI. Let's learn the essentials!

### 2.1 Variables & Data Types

Variables store information for later use.

````python
# Variables - storing information
name = "AI Student"        # String (text)
age = 16                   # Integer (whole number)
gpa = 3.8                  # Float (decimal)
is_excited = True          # Boolean (True/False)

print(f"Hi, I'm {name}!")
print(f"I'm {age} years old with a {gpa} GPA.")
print(f"Am I excited about AI? {is_excited}!")
````

### 2.2 Lists & Loops

Lists hold multiple items. Loops repeat actions — essential for AI training!

````python
# List of AI tools we'll learn
ai_tools = ["ChatGPT", "Stable Diffusion", "Whisper", "PyTorch"]

print("🛠️ AI Tools we'll use:")
for i, tool in enumerate(ai_tools, 1):
    print(f"  {i}. {tool}")

print(f"\nTotal tools: {len(ai_tools)}")
````

### 2.3 Functions

Functions are reusable blocks of code.

````python
def greet_student(name, favorite_ai):
    """Create a personalized greeting for AI students."""
    return f"👋 Welcome, {name}! I hear you love {favorite_ai}!"

# Try it!
message = greet_student("Alex", "ChatGPT")
print(message)

# Your turn - change the name and favorite AI!
your_message = greet_student("YOUR_NAME", "YOUR_FAVORITE_AI")
print(your_message)
````

### 2.4 Conditionals

Make decisions with if/else — just like AI makes decisions!

````python
# Simple AI decision making
confidence = 0.85  # AI's confidence level (0 to 1)

if confidence > 0.9:
    print("🎯 AI is very confident!")
elif confidence > 0.7:
    print("👍 AI is fairly confident.")
elif confidence > 0.5:
    print("🤔 AI is somewhat unsure.")
else:
    print("❓ AI is guessing.")
````

## 🧠 Part 4: Core AI Concepts

Understanding these concepts makes everything else easier!

### 📊 The AI Learning Pipeline

```
DATA  →  TRAINING  →  MODEL  →  INFERENCE
 📊         🏋️         🧠          🔮

Example:
🐶🐱 Images  →  Learning  →  Classifier  →  "This is a dog!"
```

**Key Terms:**
- **Dataset**: Collection of examples (like 1000 cat & dog pictures)
- **Training**: Teaching the AI patterns from data
- **Model**: The "brain" that stores learned knowledge
- **Inference**: Using the trained model to make predictions

### 🎓 Training vs Inference

| Stage | Description | Analogy |
|:--|:--|:--|
| **Training** | AI learns from examples | 📚 Studying for a test |
| **Inference** | AI applies knowledge | 📝 Taking the test |

We mostly do **inference** (using pre-trained models) in this course!

## ⚖️ Part 6: AI Ethics 101

With great AI power comes great responsibility!

### 🚨 Key Ethical Considerations

| Issue | Description | Example |
|:--|:--|:--|
| **Bias** | AI can inherit human biases | Facial recognition works worse on certain groups |
| **Hallucinations** | AI makes up false info | ChatGPT citing fake research papers |
| **Privacy** | Training on personal data | AI trained on your photos without consent |
| **Deepfakes** | Realistic fake content | Fake videos of real people |
| **Job Impact** | Automation of tasks | AI replacing certain jobs |

### ✅ Responsible AI Use
- Always verify AI-generated information
- Don't use AI to harm or deceive others
- Respect copyright and consent
- Be transparent about AI use
- Consider the impact on others

````python
# Challenge 2: AI Confidence Simulator
# Create a function that simulates AI making decisions

import random

def ai_decision(question):
    """Simulate an AI making a decision with confidence."""
    confidence = random.uniform(0.4, 0.99)
    answers = ["Yes", "No", "Maybe", "Probably", "Unlikely"]
    answer = random.choice(answers)
    
    print(f"❓ Question: {question}")
    print(f"🤖 AI Answer: {answer}")
    print(f"📊 Confidence: {confidence:.1%}")
    
# Try it!
ai_decision("Will it rain today?")
print()
ai_decision("Should I study AI?")
````
