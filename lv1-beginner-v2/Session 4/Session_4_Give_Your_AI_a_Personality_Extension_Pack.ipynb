## Challenge 3 — Tiny Sentiment Checker
This is a simple way to react to happy or sad player messages.

```python
def simple_sentiment(text):
    text = text.lower()
    happy_words = ["good", "great", "awesome", "yay", "fun"]
    sad_words = ["bad", "hard", "sad", "boring", "upset"]

    happy_score = sum(word in text for word in happy_words)
    sad_score = sum(word in text for word in sad_words)

    if happy_score > sad_score:
        return "positive"
    if sad_score > happy_score:
        return "negative"
    return "neutral"

for msg in ["This is awesome!", "This is hard", "okay"]:
    print(msg, "->", simple_sentiment(msg))
```
