## 🧠 Big Idea

You are not learning pieces anymore.

👉 You are building a **complete system**:

Image → Guess → AI → Score → Repeat


````python
# 🧩 Step 1: Game Setup

images = ["cat", "dog", "car", "banana"]  # placeholder labels
score = 0
rounds = len(images)

print("🎮 Welcome to your AI Guessing Game!")
````

## 🧪 Step 2: Basic AI Response (Starter)

This is a simple version — we will improve it next session.


````python
def ai_response(user_answer, correct_label):
    if user_answer.lower() == correct_label.lower():
        return "✅ Correct!"
    else:
        return f"❌ Wrong! It was {correct_label}."
````

## 🎮 Step 3: Game Loop

This is the core of your game.


````python
for answer in images:
    print("\n🖼️ Guess the image!")
    user_input = input("Your guess: ")

    response = ai_response(user_input, answer)
    print(response)

    if user_input.lower() == answer:
        score += 1

print(f"\n🏆 Final Score: {score}/{rounds}")
````
