## 🎯 The Big Picture

### How AI Creates Images

```
📝 Your Text Prompt    →    🧠 Diffusion Model    →    🎨 Generated Image
"a cat wearing a hat"        (noise → image)          [beautiful image!]
```

AI image generation uses **Diffusion Models** - they start with random noise and gradually "paint" your image based on your text description.

<details>
<summary><strong>🔬 How Do Diffusion Models Work? (Click to Expand)</strong></summary>

### The Diffusion Process

1. **Forward Process (Training)**: Add noise to images until they're pure static
2. **Reverse Process (Generation)**: Learn to remove noise step by step

```
Step 1: 📺 Pure noise (random pixels)
Step 2: 🌫️ Some shapes emerge
Step 3: 🎭 Details appear
Step 4: 🎨 Final image!
```

The text prompt "guides" which direction to denoise → creating matching images!

</details>

## ⚙️ Part 1: Setup

````python
import torch
import matplotlib.pyplot as plt
from diffusers import AutoPipelineForText2Image, AutoPipelineForImage2Image
from PIL import Image
import os

# Check for GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"✅ Using device: {device}")

if device == "cuda":
    print(f"🎮 GPU: {torch.cuda.get_device_name(0)}")
````

````python
# Load the model (SD Turbo - fast and efficient!)
print("Loading model... (this may take a minute)")

pipe = AutoPipelineForText2Image.from_pretrained(
    "stabilityai/sd-turbo",
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
).to(device)

print("🎨 Model ready!")
````

---

## 🖼️ Part 2: Your First Image

````python
def generate_image(prompt, num_steps=4):
    """Generate an image from a text prompt."""
    image = pipe(prompt, num_inference_steps=num_steps, guidance_scale=0.0).images[0]
    return image

def show_image(image, title="AI Generated Image"):
    """Display an image with a title."""
    plt.figure(figsize=(8, 8))
    plt.imshow(image)
    plt.axis('off')
    plt.title(title)
    plt.show()
````

````python
# Generate your first image!
prompt = "a cute robot painting a sunset, digital art style"
image = generate_image(prompt)
show_image(image, "🎨 My First AI Image!")
````

### 🎮 Try Your Own Prompts!

````python
# Experiment with different prompts!
my_prompts = [
    "a magical forest with glowing mushrooms, fantasy art",
    "an astronaut playing basketball on the moon, digital illustration",
    "a cozy coffee shop in a rainy city, anime style"
]

for prompt in my_prompts:
    print(f"\n🎨 Generating: {prompt}")
    image = generate_image(prompt)
    show_image(image, prompt[:50] + "...")
````

## ✍️ Part 3: Advanced Prompting

### The Prompt Recipe for Images

```
SUBJECT + STYLE + QUALITY + LIGHTING + DETAILS + NEGATIVE
```

| Component | Example | Effect |
|:--|:--|:--|
| **Subject** | "a wise owl wizard" | What to generate |
| **Style** | "studio ghibli style" | Artistic look |
| **Quality** | "8k, highly detailed" | Resolution/detail |
| **Lighting** | "golden hour lighting" | Mood/atmosphere |
| **Details** | "wearing spectacles" | Specific features |
| **Negative** | "blurry, distorted" | What to avoid |

````python
# Comparison: Simple vs Detailed prompts

# Simple prompt
simple_prompt = "a dragon"

# Detailed prompt
detailed_prompt = """a majestic dragon with iridescent scales,
perched on a crystal mountain peak, fantasy art style,
dramatic lighting, highly detailed, 8k resolution,
chromatic aberration, volumetric fog"""

print("📌 Simple prompt:")
image1 = generate_image(simple_prompt)
show_image(image1, "Simple: 'a dragon'")

print("📌 Detailed prompt:")
image2 = generate_image(detailed_prompt)
show_image(image2, "Detailed prompt")
````

### 🎭 Style Keywords Reference

| Category | Keywords |
|:--|:--|
| **Art Styles** | oil painting, watercolor, pixel art, anime, photorealistic, pencil sketch |
| **Lighting** | golden hour, dramatic lighting, neon glow, soft lighting, backlit |
| **Quality** | 8k, highly detailed, professional, masterpiece, award-winning |
| **Mood** | serene, dramatic, mysterious, cheerful, dark, whimsical |
| **Visual Effects** | bokeh, depth of field, lens flare, motion blur, chromatic aberration |

````python
# Same subject, different styles!

subject = "a cat sitting on a windowsill"
styles = [
    "anime style, studio ghibli",
    "oil painting, classical, rembrandt lighting",
    "pixel art, 16-bit, retro game style",
    "cyberpunk, neon lights, futuristic"
]

print("🎨 Same subject, different styles:\n")
for style in styles:
    prompt = f"{subject}, {style}"
    print(f"Style: {style}")
    image = generate_image(prompt)
    show_image(image, style)
````

## 🧩 Part 4: Negative Prompts (Optional)

**Negative prompts** tell the AI what NOT to include.

````python
def generate_with_negative(prompt, negative_prompt="", steps=4):
    """Generate image with negative prompt support."""
    # Note: SD Turbo doesn't support negative prompts directly
    # This is for demonstration - would work with standard SD
    full_prompt = prompt
    if negative_prompt:
        print(f"📝 Prompt: {prompt}")
        print(f"❌ Avoid: {negative_prompt}")
    
    image = pipe(full_prompt, num_inference_steps=steps, guidance_scale=0.0).images[0]
    return image

# Example
prompt = "a portrait of a friendly robot, digital art"
negative = "scary, damaged, broken, rusty, dark"

image = generate_with_negative(prompt, negative)
show_image(image, "Friendly Robot")
````

### Common Negative Prompts

| For | Negative Words |
|:--|:--|
| **Quality** | blurry, low quality, distorted, ugly, deformed |
| **People** | extra fingers, bad anatomy, cloned face |
| **Style** | cartoon (if you want realism), plain, boring |
| **Content** | violent, scary, nsfw |

## 🧪 Part 6: Image Editing (Optional Bonus)

If you have time, let's explore image-to-image transformation!

````python
# Load image-to-image pipeline
try:
    pipe_i2i = AutoPipelineForImage2Image.from_pipe(pipe)
    print("✅ Image-to-Image pipeline ready!")
except Exception as e:
    print(f"⚠️ Image-to-image not available: {e}")
````

````python
def transform_image(input_image, prompt, strength=0.5):
    """Transform an existing image based on a prompt.
    
    strength: 0.0 = very similar to input, 1.0 = very different
    """
    # Resize to standard size
    input_image = input_image.resize((512, 512))
    
    result = pipe_i2i(
        prompt=prompt,
        image=input_image,
        strength=strength,
        num_inference_steps=4,
        guidance_scale=0.0
    ).images[0]
    
    return result

# Example: Transform one of our game assets
# original = Image.open("game_assets/hero.png")
# transformed = transform_image(original, "the same robot but in winter clothing, snowy")
# show_image(transformed, "Transformed Hero")
````
