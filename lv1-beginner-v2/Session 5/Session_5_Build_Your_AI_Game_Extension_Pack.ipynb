## Challenge 2 — Difficulty Levels
Give the player different score rules.

```python
difficulty = "normal"   # easy, normal, hard

def score_change(correct, difficulty_name):
    if difficulty_name == "easy":
        return 10 if correct else -2
    if difficulty_name == "hard":
        return 15 if correct else -8
    return 10 if correct else -5

for result in [True, False]:
    print(result, "->", score_change(result, difficulty))
```
