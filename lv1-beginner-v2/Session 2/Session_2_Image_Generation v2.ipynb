```python
# Imports
from openai import OpenAI

from pathlib import Path
from PIL import Image
from io import BytesIO
import matplotlib.pyplot as plt
import os
import base64

from getpass import getpass
```

```python
# Enter your OpenAI API key
# Tip: paste carefully; nothing will show while typing.
api_key = getpass("Enter your OpenAI API key: ")

# --- ADDED FOR DEBUGGING: Print a snippet of the key to confirm it was captured ---
print(f"API key captured (first 5 chars): {api_key[:5]}...")
# -----------------------------------------------------------------------------------

client = OpenAI(api_key=api_key)
print("API key loaded.")
```

## Helper Functions

Run the next cell once. It gives you simple tools to:
- generate an image
- preview it
- save it into a folder


```python
OUTPUT_DIR = Path("generated_images")
OUTPUT_DIR.mkdir(exist_ok=True)

def show_image(img, title=None):
    plt.figure(figsize=(6, 6))
    plt.imshow(img)
    plt.axis("off")
    if title:
        plt.title(title)
    plt.show()

def generate_image(
    prompt,
    size="1024x1024",
    save_folder="generated_images",
    filename=None,
    background="auto",
):
    response = client.images.generate(
        model="gpt-image-1",
        prompt=prompt,
        size=size,
        background=background,
    )

    image_b64 = response.data[0].b64_json
    image_bytes = base64.b64decode(image_b64)
    img = Image.open(BytesIO(image_bytes))

    save_path = None
    if filename:
        folder = Path(save_folder)
        folder.mkdir(parents=True, exist_ok=True)
        save_path = folder / filename
        img.save(save_path)

    return img, save_path

def safe_name(text):
    cleaned = "".join(ch.lower() if ch.isalnum() else "_" for ch in text).strip("_")
    while "__" in cleaned:
        cleaned = cleaned.replace("__", "_")
    return cleaned[:80] or "image"

print("Helpers ready.")
```

Now make your own image.

## Part 3 — Style Experiments

Keep the same subject, but change the style.

Example styles:
- cartoon style
- watercolor painting
- pixel art
- realistic photo
- anime style


```python
subject = "a small dog wearing sunglasses"

styles = [
    "cartoon style",
    "watercolor painting",
    "pixel art",
    "realistic photo"
]

for style in styles:
    prompt = f"{subject}, {style}, clean composition, highly detailed"
    filename = f"style_{safe_name(style)}.png"
    img, path = generate_image(prompt, filename=filename)
    show_image(img, title=style)
    print("Saved to:", path)
```

## Part 4 — Build a Small Dataset

For later classes, it helps to organize images into categories.

Example categories:
- cat
- dog
- robot
- food

Try generating **2–4 images per category**.


```python
DATASET_DIR = Path("game_assets")
DATASET_DIR.mkdir(exist_ok=True)

categories = {
    "cat": [
        "a cute cat, cartoon style, bright colors",
        "a fluffy cat, cartoon style, sitting on a chair"
    ],
    "dog": [
        "a happy dog, cartoon style, bright colors",
        "a playful dog, cartoon style, running in a park"
    ],
    "robot": [
        "a friendly robot, cartoon style, simple background",
        "a smiling robot, cartoon style, waving"
    ],
}

for label, prompts in categories.items():
    label_dir = DATASET_DIR / label
    label_dir.mkdir(parents=True, exist_ok=True)

    for idx, prompt in enumerate(prompts, start=1):
        filename = f"{label}_{idx}.png"
        img, path = generate_image(prompt, save_folder=label_dir, filename=filename)
        print(f"Saved {label}: {path}")
```

### Why this matters

A cleaner dataset makes later AI tasks easier:
- better labels
- less confusion
- easier training and testing
