## Challenge 2 — Top-2 Guesses
Sometimes the AI is choosing between two close answers.

```python
def top_two_guesses(image_path):
    image = Image.open(image_path).convert("RGB")
    x = transform(image).unsqueeze(0)

    with torch.no_grad():
        logits = model(x)
        probs = torch.softmax(logits, dim=1)[0]

    pairs = list(zip(classes, probs.tolist()))
    pairs = sorted(pairs, key=lambda x: x[1], reverse=True)
    return pairs[:2]

if image_paths:
    p = random.choice(image_paths)
    show_image(p)
    print(top_two_guesses(p))
```
