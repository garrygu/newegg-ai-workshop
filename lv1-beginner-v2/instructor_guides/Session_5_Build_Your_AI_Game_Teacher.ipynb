- `cat_01.png` → answer is `cat`
- `dog_02.png` → answer is `dog`

This keeps the final session stable and fast.

## Live Demo Plan

## Teaching Tips

### Draw the flow
Write this on the board or projector:

Show image  
→ player guesses  
→ game checks answer  
→ score changes  
→ AI reacts

### Keep loops concrete
Use the phrase **“repeat the same steps each round”** instead of only saying “for loop.”

### Pair playtesting works well
Students often catch confusing messages faster when someone else plays their game.

## Common Problems + Fixes

### 1) Images do not match labels
Rename files to include the label text.

### 2) Students struggle with the loop
First run one round manually. Then show how a loop repeats that pattern.

### 3) Students finish early
Use the extension notebook for hints, difficulty, agent mode, or scoreboards.

## Assessment Ideas
Students can demonstrate success if they:
- run a complete round
- update score correctly
- show at least one AI reaction
- finish a full mini-game
- explain how Sessions 3–5 connect together
