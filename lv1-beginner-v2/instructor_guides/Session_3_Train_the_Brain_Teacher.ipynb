# 👩‍🏫 Session 3 — Train the Brain of Your AI (Teacher)

This notebook is the instructor version for a 2-hour class.

## Lesson Goal
Students understand that:
- an image classifier is the **brain** of the future game
- the model predicts labels from images
- AI confidence is not the same as certainty
- mistakes are useful for learning

## Recommended Timing
- 0–10 min: welcome + project storyline
- 10–20 min: live demo
- 20–55 min: guided setup and model load
- 55–80 min: student image testing
- 80–95 min: discussion of errors
- 95–110 min: save settings + recap
- 110–120 min: extensions or buffer

## Prep Checklist
Before class:
- confirm `models/classifier.pt` exists
- confirm `game_assets/` contains example images
- run the notebook once on the teaching machine
- prepare a backup folder of 6–10 sample images
- decide the final class labels in advance

```python
# Teacher quick check
from pathlib import Path

print("Model exists:", Path("models/classifier.pt").exists())
print("Assets folder exists:", Path("game_assets").exists())

if Path("game_assets").exists():
    files = []
    for ext in ["*.png", "*.jpg", "*.jpeg", "*.webp"]:
        files.extend(Path("game_assets").glob(ext))
    print("Image count:", len(files))
```

## Live Demo Plan (first 10–20 minutes)

1. Show one image  
2. Ask students to guess first  
3. Run the AI prediction  
4. Reveal the AI answer  
5. Ask: “Why might the AI be wrong sometimes?”

Keep this part short and exciting.

```python
# Optional demo cell
import random
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

demo_images = []
for ext in ["*.png", "*.jpg", "*.jpeg", "*.webp"]:
    demo_images.extend(Path("game_assets").glob(ext))

if demo_images:
    p = random.choice(sorted(demo_images))
    print("Demo image:", p.name)
    img = Image.open(p).convert("RGB")
    plt.figure(figsize=(4,4))
    plt.imshow(img)
    plt.axis("off")
    plt.show()
else:
    print("Add demo images to game_assets/")
```

## Teaching Notes by Section

### Imports / setup
Keep students together here. Do not let the room spread too far in pace.

### Labels
This is a good moment to connect to their own project theme.

### Model loading
Explain: “The model is already trained. Today we are testing and using it.”

### Prediction
Use the word **guess** often. It feels intuitive for this age group.

## Common Problems + Fixes

### 1) `FileNotFoundError` for model
Use a prepared backup model, or switch to a teacher demo mode.

### 2) No images in `game_assets`
Provide a zip or a class-shared folder.

### 3) Students stuck on confidence
Say: “Confidence is how strongly the AI leans toward one answer.”

### 4) Students finish early
Use the extension notebook.

## Discussion Prompts

Use 2–3 of these:
- Which image was easiest for the AI?
- Which image confused it most?
- Can AI be confident and still wrong?
- What kind of training images would improve the model?

## Assessment Ideas

Students can demonstrate success if they:
- run a prediction on at least one image
- explain what a label is
- describe one case where AI can fail
- save settings for future classes
