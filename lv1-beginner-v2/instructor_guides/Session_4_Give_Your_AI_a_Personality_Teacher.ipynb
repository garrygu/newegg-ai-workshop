## Teaching Tips

### Keep it game-based
Always tie responses to gameplay:
- welcome
- correct answer
- wrong answer
- hint
- final score

### Encourage expressive edits
Students can add:
- emojis
- tone
- catchphrases

### Keep the scope simple
Voice should be optional or teacher-demo only.
Do not let voice setup consume the class.

## Common Problems + Fixes

### 1) Students keep breaking strings
Remind them to keep quotes balanced.

### 2) Students want too much complexity
Guide them back to 3 core reactions:
- welcome
- correct
- wrong

### 3) Students finish early
Use the extension notebook for mood, sentiment, or reaction variety.

## Discussion Prompts
- What kind of personality makes a game more fun?
- Can the same AI brain feel different with different words?
- Why does wording matter in technology products?

## Assessment Ideas

Students can demonstrate success if they:
- choose a personality
- make one “correct” and one “wrong” response
- explain how their AI feels different from someone else’s
- save their settings for Session 5
