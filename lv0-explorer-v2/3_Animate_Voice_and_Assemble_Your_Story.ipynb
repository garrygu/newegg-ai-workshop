# PART 1: AI Video Animation 🎬 {#video}

## 🎥 What is AI Video Animation?

AI video animation turns your **still images into moving videos**!

**What it can do:**
- Make characters move or gesture
- Add camera movements (zoom, pan)
- Create smooth transitions
- Bring static scenes to life!

**Example:**
- Still image of a robot → Video of robot waving its hand!

---

## 🛠️ Tool: Runway ML

[👉 Open Runway ML](https://app.runwayml.com)

**Why Runway ML?**
- ✅ Easy to use
- ✅ Great quality animations
- ✅ Free tier available
- ✅ Fast generation (30-60 seconds)
- ✅ Animate YOUR specific images from Class 2!

# PART 2: Assemble Your Story in Canva 🎨 {#assemble}

## 🎬 It's Assembly Time!

Now we'll put everything together IN ONE TOOL - Canva has everything you need!

**What you have:**
- ✅ Images from Class 2
- ✅ Videos from Part 1 (if you made them)
- ✅ Your storyboard with scripts

**What Canva will do:**
- 📦 Combine all your visuals
- 🎤 Generate AI voices from your script (built-in!)
- 🎵 Add music (optional)
- 🎬 Create your final video!

---

## 🛠️ Tool: Canva Video Editor

[👉 Open Canva](https://www.canva.com)

**Why Canva?**
- ✅ Super easy to use
- ✅ Drag and drop interface
- ✅ Free version works great
- ✅ **Built-in AI text-to-speech** (no extra tool needed!)
- ✅ Can combine images, videos, and audio
- ✅ Professional-looking results

---

## 📚 Step-by-Step Assembly Guide

### **Step 1: Create New Video Project**

1. Go to [Canva](https://www.canva.com)
2. Click **"Create a design"**
3. Select **"Video"**
4. Choose **"1920 x 1080 px"** (standard HD)
5. Your blank canvas is ready!

**Or:** Search for "Video Presentation" templates if you want a starting point!

---

### **Step 2: Upload Your Visual Assets**

1. Click **"Uploads"** in the left sidebar
2. Click **"Upload files"**
3. Upload your images and videos:
   - All images from `my_ai_story/images/`
   - All videos from `my_ai_story/videos/` (if you made any)
4. Wait for everything to upload

**Pro Tip:** You can drag and drop files directly!

**Note:** You DON'T need to upload voice files - Canva will create them!

---

### **Step 3: Add Pages for Each Scene**

Each scene in your storyboard = 1 page in Canva!

1. Look at how many scenes you have (3-5 scenes)
2. Add pages: Click **"+" button** at the bottom
3. Create one page per scene

**Example:**
- Page 1: Title card (optional)
- Page 2: Scene 1
- Page 3: Scene 2
- Page 4: Scene 3
- Page 5: Ending/credits (optional)

---

### **Step 4: Build Scene 1 with Voice!**

Let's do the first scene together step-by-step:

**4a. Add Visual**
1. Go to Page 1 (or 2 if you have a title)
2. From Uploads, drag your Scene 1 image OR video
3. Resize to fill the page

---

**4b. Add AI Voice (Canva's Text-to-Speech!) 🎤**

This is the cool part - Canva will generate voices for you!

1. Add a text box: Click **"Text"** in the left sidebar
2. Click **"Add a text box"**
3. Type or paste your Scene 1 script from your storyboard
   - Example: "Hello! My name is Robo, and I'm here to help you!"
4. Select the text you just typed
5. Click **"More"** (three dots) in the text toolbar
6. Select **"Text to speech"** or **"Voice over"**
7. Choose a voice:
   - Try different voices by clicking on them
   - Pick one that fits your character!
   - Options include: Narrator, Friendly, Young, Energetic, etc.
8. Click **"Generate"**
9. The AI voice is automatically added to your page!

**🎯 Pro Tip:** You can hide the text box after generating voice (just make it tiny or delete it)

---

**4c. Set Duration**
1. Click on the page thumbnail at bottom
2. Click the timer icon
3. Set duration based on voice length
   - Listen to the voice first to see how long it is
   - Add 1-2 seconds buffer
   - Example: If voice is 8 seconds → set page to 9-10 seconds

---

**4d. Preview Scene 1**
1. Click the **Play button** (▶️) at top right
2. Watch/listen to Scene 1
3. Does it work? Great!
4. Need to adjust? Change the timing or regenerate voice

---

### **Step 5: Build Remaining Scenes**

Repeat Step 4 for each of your scenes!

**For Scene 2:**
1. Add Scene 2 visual (image or video)
2. Add text with Scene 2 script
3. Generate AI voice with text-to-speech
4. Set duration to match voice
5. Preview

**For Scene 3:**
1. Add Scene 3 visual
2. Add text with Scene 3 script
3. Generate AI voice
4. Set duration
5. Preview

**Continue for all scenes!**

---

### **Step 6: Add Transitions (Optional)**

Make scenes flow smoothly!

1. Click between two pages (the gap)
2. Click **"Transition"**
3. Choose a transition style:
   - **Fade** - Smooth and professional ⭐ Recommended!
   - **Slide** - Dynamic movement
   - **Circle wipe** - Fun and creative

**Recommendation:** Use the same transition throughout for consistency!

**Duration:** 0.5 - 1 second is perfect

---

### **Step 7: Add Polish ✨ (Optional)**

**Optional Enhancements:**

**A. Title Card (Beginning)**
1. Add a page at the start
2. Add text with your story title
3. Make it look nice with colors/fonts!
4. Duration: 3-4 seconds
5. Optional: Add voice saying the title!

**B. Text Captions**
- Add text to show what's being said (like subtitles)
- Keep them short and readable!
- Place at bottom of screen

**C. Background Music (Optional)**\n\n**If you created music in Class 2:**\n1. Click **"Uploads"** in Canva\n2. Upload your music file from `my_ai_story/music/`\n3. Drag it onto your timeline\n4. **IMPORTANT:** Lower volume to 20-30%!\n\n**OR use Canva's music library:**\n1. Click **"Audio"** in sidebar\n2. Search Canva's music library\n3. Try keywords: "happy", "adventure", "inspiring"\n4. Choose something that fits your mood\n5. **IMPORTANT:** Lower the music volume to 20-30%!\n\n**D. Visual Effects**
- Click on an image
- Click **"Animate"**
- Choose subtle effects (breathe, pan, zoom)
- Don't overdo it!

**⚠️ Simple is often better!** Don't add too many effects.

---

### **Step 8: Preview Full Story**

Time to watch your complete story!

1. Click the **Play button** (▶️) at top right
2. Watch your ENTIRE story from start to finish
3. Check:
   - □ Does it flow well?
   - □ Is the audio clear?
   - □ Can you hear voices over music?
   - □ Are the timings right?
   - □ Does it tell your story?
   - □ Total length is 1-2 minutes?

**Make adjustments as needed!**

---

### **Step 9: Export Your Final Story! 🎉**

1. Click **"Share"** (top right)
2. Click **"Download"**
3. Select **"MP4 Video"**
4. Quality: **Standard** (free) or **High** (with Canva Pro)
5. Click **"Download"**
6. Wait for it to render (may take 2-5 minutes)
7. Save to your computer!

**✅ Save as:** `my_ai_story_final.mp4`

**🎊 CONGRATULATIONS! Your story is complete!**
